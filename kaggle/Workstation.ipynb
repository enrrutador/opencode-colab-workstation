# OpenCode Cloud Workstation — Kaggle

Workstation **persistente y reanudable** para OpenCode sobre un runtime Kaggle **efímero**.

### Antes de ejecutar

1. Crea un Kaggle Dataset **privado**, p. ej. `tu-usuario/opencode-workstation-persistence`
2. En **Settings → Secrets** del notebook:
   - `NVIDIA_API_KEY` (requerido)
   - `OPENCODE_CLOUD_DATASET` = `tu-usuario/opencode-workstation-persistence`
   - `GITHUB_TOKEN` / `GITHUB_REPO` (opcional, solo versionado de código)
3. Ejecuta las celdas en orden.

### Qué hace el bootstrap

1. Descarga el Dataset (`kagglehub`) o inicia workstation fresca
2. Restaura workspace / estado / config
3. Asegura Node + OpenCode (idempotente)
4. Configura NVIDIA NIM solo vía env (`{env:NVIDIA_API_KEY}`)
5. Arranca OpenCode Web en `0.0.0.0:4096` **local al runtime**
6. Watchdog con restart real (no publica Dataset)

> **Nota:** Kaggle no expone URL pública oficial para ese puerto. OpenCode Web es local al kernel.
