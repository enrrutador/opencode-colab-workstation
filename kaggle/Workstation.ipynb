# OpenCode Cloud Workstation — Kaggle

Workstation persistente y reanudable para OpenCode sobre un runtime Kaggle efímero.

**Antes de ejecutar:**
1. Crea un Kaggle Dataset privado `owner/opencode-workstation-persistence`
2. En **Kaggle Secrets** define:
   - `NVIDIA_API_KEY`
   - `OPENCODE_CLOUD_DATASET` = `owner/opencode-workstation-persistence`
   - `GITHUB_REPO` / `GITHUB_TOKEN` (opcional, solo versionado de código)
3. Ejecuta las celdas en orden.

OpenCode Web queda en `http://127.0.0.1:4096` **dentro del runtime**. Kaggle no expone URL pública oficial para ese puerto.
